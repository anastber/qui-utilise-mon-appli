```python
import pandas as pd
```

```python
def read_ds(ds_name: str):
    with open(f"data/{ds_name}.csv") as f:
        max_fields = max(len(line.rstrip("\n").split(",")) for line in f)
        f.seek(0)
        names = ["util", "navigateur"] if ds_name == "train" else ["navigateur"]
        names.extend(range(max_fields - len(names)))
        return pd.read_csv(f, names=names, dtype=str)

features_test = read_ds("test")
```
